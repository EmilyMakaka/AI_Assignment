```python
import heapq

GRAPH = {
    "Nairobi":  {"Naivasha": 92, "Narok": 145},
    "Naivasha": {"Nakuru": 64},
    "Nakuru":   {"Kericho": 100, "Eldoret": 156},
    "Narok":    {"Bomet": 100},
    "Bomet":    {"Kericho": 60, "Kisii": 80},
    "Kericho":  {"Kisumu": 100},
    "Eldoret":  {"Kakamega": 80},
    "Kakamega": {"Kisumu": 50},
    "Kisii":    {"Kisumu": 115},
    "Kisumu":   {},
}

START, GOAL = "Nairobi", "Kisumu"


def fmt_frontier(frontier):
    return "[" + ", ".join(f"({n}, {c})" for c, n, _ in sorted(frontier)) + "]"


def uniform_cost_search(graph, start, goal):
    frontier = [(0, start, [start])]
    explored = set()
    best_cost = {start: 0}
    step = 0

    print(f"Uniform Cost Search: {start} -> {goal}\n" + "=" * 60)
    while frontier:
        step += 1
        print(f"\nStep {step}")
        print(f"  Frontier : {fmt_frontier(frontier)}")
        cost, node, path = heapq.heappop(frontier)

        if node in explored:
            print(f"  Skipping {node} (already explored with lower cost)")
            continue

        print(f"  Expand   : {node}  (cost so far = {cost})")
        print(f"  Path     : {' -> '.join(path)}")

        if node == goal:
            print("  Goal test: GOAL REACHED (tested when popped, so optimal)")
            print(f"  Explored : {sorted(explored | {node})}")
            return cost, path

        explored.add(node)
        for nbr, w in graph[node].items():
            new_cost = cost + w
            if nbr in explored:
                continue
            if nbr not in best_cost or new_cost < best_cost[nbr]:
                best_cost[nbr] = new_cost
                heapq.heappush(frontier, (new_cost, nbr, path + [nbr]))
                print(f"    add/update {nbr}: {cost} + {w} = {new_cost}")
            else:
                print(f"    ignore {nbr}: {new_cost} >= best known {best_cost[nbr]}")
        print(f"  Explored : {sorted(explored)}")

    return None, None


if __name__ == "__main__":
    total, route = uniform_cost_search(GRAPH, START, GOAL)
    print("\n" + "=" * 60)
    if route:
        print("OPTIMAL PATH :", " -> ".join(route))
        print("TOTAL COST   :", total, "km")
    else:
        print("No path found.")
```
